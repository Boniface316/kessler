```python
file_name = "data/train_data.csv"
number_of_events = 100

event_idx = 0
events_original = kelvins_to_event_dataset(
    file_name, drop_features=["c_rcs_estimate", "t_rcs_estimate"], num_events=100
)

event_original = events_original._events[event_idx]
event_original_df = event_original.to_dataframe()

cols_to_keep = dir(InputsSchema)

events = CSVReader(path=file_name, number_of_events=100).read(columns_to_keep=cols_to_keep)
event0 = events.events[event_idx]
```

```python
cdm_to_add = cdm0 = events.events[event_idx].cdms[1]

event_copy = event0.copy()
assert event_copy == event0

event_copy.add(cdm_to_add)

assert len(event_copy) == 6
assert event_copy.cdms[1] == cdm_to_add
# expect this to fail
assert event_copy.cdms[5] == events.events[1].cdms[0]
```

```python
event_copy = event0.copy()
assert event_copy == event0
```
