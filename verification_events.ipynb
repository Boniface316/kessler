```python
events_df = events.to_dataframe()
events_original_df = events_original.to_dataframe()


events_df = events_df.drop(columns=["EVENT_ID"])
events_original_df = events_original_df.rename(columns=column_mapping)

events_original_df = events_original_df.reindex(columns=events_df.columns)
columns_to_avoid = [
    "CCSDS_CDM_VERS",
    "CREATION_DATE",
    "ORIGINATOR",
    "MESSAGE_FOR",
    "MESSAGE_ID",
    "TCA",
    "t_OBJECT",
    "t_CD_AREA_OVER_MASS",
    "t_CR_AREA_OVER_MASS",
    "t_TIME_LASTOB_START",
    "t_TIME_LASTOB_END",
    "c_OBJECT",
    "c_CD_AREA_OVER_MASS",
    "c_CR_AREA_OVER_MASS",
    "c_TIME_LASTOB_START",
    "c_TIME_LASTOB_END",
]

events_df = events_df.drop(columns=columns_to_avoid)
events_original_df = events_original_df.drop(columns=columns_to_avoid)
assert events_df.equals(events_original_df)
```

```python
original_common_features_list = events_original.common_features()

original_common_features_list = [
    column_mapping.get(col, col) for col in original_common_features_list
]
original_common_features_list = sorted(original_common_features_list)

features_list = events.common_features()
features_list = sorted(features_list)

for col in [
    "EVENT_ID",
    "MESSAGE_FOR",
    "MESSAGE_ID",
    "ORIGINATOR",
    "c_CD_AREA_OVER_MASS",
    "c_CR_AREA_OVER_MASS",
    "t_CD_AREA_OVER_MASS",
    "t_CR_AREA_OVER_MASS",
]:
    if col in features_list:
        features_list.remove(col)

assert original_common_features_list == features_list, (
    "Common features do not match between original and modified events."
)
```

```python
assert len(events.get_CDMs()) == len(events_original.get_CDMs()), (
    "Number of CDMs do not match between original and modified events."
)
```

```python
assert len(events_original.filter(lambda event: len(event) > 4)) == len(
    events.filter(lambda event: len(event) > 4)
)
```
